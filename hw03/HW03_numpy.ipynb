````python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
````

#### 4 Маркетинг предлагает скидку 15% на оплаченные заказы дороже 50 000 ₸. Сколько выручки магазин недополучит? Посчитайте дважды: со всеми заказами и без заказов клиента TEST. Почему ответы различаются в несколько раз?

````python
# Стандарт
discount_price = (amounts[paids & (amounts>50000)]*0.85).sum() + amounts[paids & (amounts<50000)].sum()
lost_revenue = paid_amounts.sum() - discount_price
print('Выручка без заказов клиента тест: ',float(lost_revenue))
````

````python
discount_price_test = (amount[paid & (amount>50000)]*0.85).sum() + amount[paid & (amount<50000)].sum()
lost_revenue_test = amount[(paid==True)].sum() - discount_price_test
print('Выручка со всех заказов: ',float(lost_revenue_test))
````

### 5 Для каждого города — доля заказов, доставленных за 1–2 дня. Список городов: np.unique(city); цикл по 6 городам здесь можно.

````python
fast_delivery = (days==1) | (days == 2)

for c in np.unique(city):
    in_city = city == c 
    total = in_city.sum()
    fast = (fast_delivery & in_city).sum()
    share = fast/total*100
    print(f"{c:<12}| {total:>8} | {fast:>8} | {share:>5.1f}%")
````
