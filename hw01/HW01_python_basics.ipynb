````python
order_count = len(amounts_int)
print(f"общее число заказов: {order_count}")
````

````python
avg_order=round(summa/order_count)
print(f"средний чек: {avg_order}")
````

Сколько заказов не оплачено и на какую сумму?

````python
unpaid_orders=0
sum_unpaid_orders=0
for a,p in zip(amounts_int,paid):
    if p == False:
        unpaid_orders += 1
        sum_unpaid_orders+=a
print(f"Количество не оплаченных заказов: {unpaid_orders}")
print(f"Cумма не оплаченных заказов: {sum_unpaid_orders}")
````

**Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью while посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?**


````python
total,month = 0,0
increase = 0
while total < 400000:
    total += 15000
    month += 1
    if month >= 2:
        increase += 1000
        total+=increase
print(total, month)
````
