## [База] 2. Типичный заказ
Для оплаченных заказов без TEST: среднее, медиана, 25-й и 75-й перцентили суммы.
Какой чек «типичный»? Что честнее показать в отчёте, среднее или медиану, и почему?

````python
amounts = df.loc[paid & not_test, "amount"]

print(f"Среднее:{amounts.mean():>10.2f}")
print(f"Медиана:{amounts.median():>10.2f}")
print(f"25-й перцентиль:{amounts.quantile(0.25):>10.2f}")
print(f"75-й перцентиль:{amounts.quantile(0.75):>10.2f}")
````

##### Вывод
Среднее (14672) на 42% больше медианы (10300). По перцентилям видно, что распределение несимметрично: от медианы до 75-го перцентиля разница 8390, а от 25-го перцентиля до медианы — всего 4420. Верхняя половина чеков разбросана почти вдвое шире, то есть среднее увеличивают отдельные крупные заказы.
Поэтому как типичный чек используем медиану — 10300; половина оплаченных заказов укладывается в 5880–18690

Доля отменённых заказов: по всем заказам, в Шымкенте, в Алматы (все написания каждого города через `isin`).
Какой город выделяется и во сколько раз?

````python
# 3
df['city'] = df['city'].str.strip().str.capitalize()
cities = df['city'].unique()
cities
````

````python
al = ['Алматы', 'Almaty']
sh = ['Шымкент', 'Shymkent']
````

````python
##### Доля отменённых заказов: по всем заказам
total_unpaid = (~paid).sum()/len(df['status'])
print(f"{'Доля отменённых заказов: по всем заказам: '} {total_unpaid:.2%}")
````

````python
unpaid_al = (df['city'].isin(al) & ~paid).sum()/(df['city'].isin(al)).sum()
unpaid_sh = (df['city'].isin(sh) & ~paid).sum()/(df['city'].isin(sh)).sum()
print(f"{'Доля отменённых заказов Алмате: '} {unpaid_al:.2%}")
print(f"{'Доля отменённых заказов Шымкенте: '} {unpaid_sh:.2%}")
````

## [Стандарт] 4. Октябрь против сентября по двум городам
Выручка оплаченных заказов без TEST за сентябрь и октябрь в Шымкенте и в Алматы, изменение в %.
Дубли пока не убираем. Где падение сильнее и во сколько раз? Что сказать менеджеру?

````python
# 4
rev_al_sep = df.loc[(paid & not_test & sep & (df['city'].isin(al))),'amount'].sum()
rev_al_oct = df.loc[(paid & not_test & octo & (df['city'].isin(al))),'amount'].sum()
print(rev_al_sep,rev_al_oct)
change_al = rev_al_oct/rev_al_sep-1
print(f"{'Изменение в городе Алматы '} {change_al:.1%}")
````

````python
rev_sh_sep = df.loc[(paid & not_test & sep & (df['city'].isin(sh))),'amount'].sum()
rev_sh_oct = df.loc[(paid & not_test & octo & (df['city'].isin(sh))),'amount'].sum()
print(rev_sh_sep,rev_sh_oct)
change_sh = rev_sh_oct/rev_sh_sep-1
print(f"{'Изменение в городе Шымкент '} {change_sh:.1%}")
````

Выручка по двум городам упала но Шымкенте упала 3 раза больше чем в Алмате. И в Шымкенте больше отмененных заказов. Предлагаю выяснить частых отмен в Шымкенте может товар оказался браковонным или не было в складе. И еще нужно что именно изменилось в Шымкенте в октябре
