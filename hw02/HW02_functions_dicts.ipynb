````python
def to_nums(amount):
    result = [i.replace(" ","") for i in amount]
    return list(map(float,result))
def clean_city(c):
    cities_clean = [a.strip().capitalize() for a in c]
    return cities_clean
````

Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.

````python
amounts = to_nums([row["amount"] for row in rows])

for i in range(len(rows)):
    rows[i]["amount"] = amounts[i]

revenue = {}
count = {}

for row in rows:
    if row["status"] == "оплачен":
        cat = row["category"]

        if cat not in revenue:
            revenue[cat] = 0
            count[cat] = 0

        revenue[cat] += row["amount"]
        count[cat] += 1

print(f"{'Категория':<15}| {'Выручка':>12} | {'Ср. чек':>10}")
for cat in revenue:
    avg = revenue[cat] / count[cat]
    print(f"{cat:<15}| {revenue[cat]:>12.0f} | {avg:>10.0f}")
````

Сколько заказов в каждом месяце? Месяц — это order_date[:7].

````python
months = {}

for row in rows:
    month = row["order_date"][:7]

    if month not in months:
        months[month] = 0

    months[month] += 1

print(f"{'Месяц':<10}| {'Заказов':>8}")
for month in sorted(months):
    print(f"{month:<10}| {months[month]:>8}")
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт
revenue = {}

for row in rows:
    if row["status"] == "оплачен":
        month = row["order_date"][:7]

        if month not in revenue:
            revenue[month] = 0

        revenue[month] += float(row["amount"])

print(f"{'Месяц':<10}| {'Выручка, ₸':>14}")
for month in sorted(revenue):
    print(f"{month:<10}| {revenue[month]:>14,.0f}")

sep = revenue["2026-09"]
oct = revenue["2026-10"]
change = (oct - sep) / sep * 100

print(f"\nОктябрь к сентябрю: {change:+.1f}%")
````
