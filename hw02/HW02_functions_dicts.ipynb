````python
def to_nums(amount):
    result = [i.replace(" ","") for i in amount]
    return list(map(int, map(float, result)))
def clean_city(c):
    cities_clean = [a.strip().capitalize() for a in c]
    return cities_clean
````

````python
amount_int = []
for i in range(len(rows)):
    amount_int.append(rows[i]['amount']) 
amount_int = to_nums(amount_int)
amount_int[0:10]
````

````python
cleaned_city = []
for i in range(len(rows)):
    cleaned_city.append(rows[i]["city"])
cleaned_city = clean_city(cleaned_city)
````

Сколько заказов в каждой категории (category)? Отсортируйте по убыванию.

````python
set(r["category"] for r in rows)
````

````python
count_order_category = {}
for row in rows:
    if row['status'] == 'оплачен':
        category = row["category"]
        count_order_category[category] = count_order_category.get(category,0)+1
sorted(count_order_category.items(), key = lambda x: x[1], reverse = True)
````

Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.

````python
revenue_by_category = {}
avg_by_category = {}
for i in range(len(rows)):
    rows[i]["amount"] = amount_int[i]

for row in rows:
    if row['status'] == 'оплачен':
        category = row["category"]
        revenue_by_category[category] = revenue_by_category.get(category,0) + row["amount"]
        avg_by_category[category] = revenue_by_category[category]/count_order_category[category]

print(f"{'Категория':<15}| {'Выручка, ₸':>14} | {'Ср. чек, ₸':>11}")
print("-" * 15 + "+" + "-" * 16 + "+" + "-" * 12)
for category in revenue_by_category:
    print(f"{category:<15}| {revenue_by_category[category]:>14,.0f} | {avg_by_category[category]:>11,.0f}")
````

### Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.

````python
revenue_by_month = {}
for row in rows:
    if row['status'] == 'оплачен':
        month = row['order_date'][:7]
        revenue_by_month[month] = revenue_by_month.get(month,0) + row["amount"]
sept = revenue_by_month['2026-09']
october = revenue_by_month['2026-10']

changes = (october - sept) / sept * 100

print(f"{'Месяц':<8}| {'Выручка':>8}")
for month in revenue_by_month:
    print(f"{month:<8}| {revenue_by_month[month]:>8,}")
    
print(f"Изменение октября к сентябрю в процентах:{changes:+.1f}%"  )
````

### Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.

````python
for i in range(len(rows)):
    rows[i]["city"] = cleaned_city[i]

canceled_order_city = {}
order_by_city = {}
for row in rows:
    city = row["city"]
    if row['status'] != 'оплачен':
        canceled_order_city[city] = canceled_order_city.get(city,0) + 1
    order_by_city[city] = order_by_city.get(city,0) + 1

for city in sorted(order_by_city):
    total = order_by_city[city]
    canceled = canceled_order_city.get(city,0)
    share = canceled / total * 100
    print(f"{city:<15}| {total:>8} | {canceled:>8} | {share:>5.1f}%")
````

Топ-5 клиентов (customer_id) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке

````python
customers= {}

for row in rows:
    if row['status'] == 'оплачен':
        customer = row['customer_id']
        customers[customer] = customers.get(customer,0)+row['amount']
top_customers = sorted(customers.items(), key=lambda x: x[1],reverse=True)
top_customers[:5]
````
